```python
churn_counts = df['Churn'].value_counts()

print("Customer Churn Distribution:")
print(churn_counts)
```

```python
churn_percentage = (
    df['Churn'].value_counts(normalize=True) * 100
)

print(churn_percentage.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Churn'
)

plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct='%1.2f%%',
    startangle=90
)

plt.title('Customer Churn Percentage')

plt.show()
```

```python
gender_churn = pd.crosstab(
    df['gender'],
    df['Churn']
)

print(gender_churn)
```

```python
gender_rate = (
    df.groupby('gender')['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(gender_rate.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=gender_rate.index,
    y=gender_rate.values
)

plt.title('Customer Churn Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
contract_rate = (
    df.groupby('Contract')['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(contract_rate.round(2))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=contract_rate.index,
    y=contract_rate.values
)

plt.title('Customer Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')

plt.xticks(rotation=15)

plt.show()
```

```python
payment_rate = (
    df.groupby('PaymentMethod')['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(payment_rate.round(2))
```

```python
plt.figure(figsize=(11, 5))

sns.barplot(
    x=payment_rate.index,
    y=payment_rate.values
)

plt.title('Customer Churn Rate by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Churn Rate (%)')

plt.xticks(rotation=25, ha='right')

plt.show()
```

```python
internet_rate = (
    df.groupby('InternetService')['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(internet_rate.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=internet_rate.index,
    y=internet_rate.values
)

plt.title('Customer Churn Rate by Internet Service')
plt.xlabel('Internet Service')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
bins = [0, 12, 24, 48, 72]

labels = [
    '0-12 Months',
    '13-24 Months',
    '25-48 Months',
    '49-72 Months'
]

df['Tenure_Group'] = pd.cut(
    df['tenure'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
tenure_rate = (
    df.groupby('Tenure_Group', observed=False)['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(tenure_rate.round(2))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=tenure_rate.index,
    y=tenure_rate.values
)

plt.title('Customer Churn Rate by Tenure')
plt.xlabel('Customer Tenure')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
partner_rate = (
    df.groupby('Partner')['Churn']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

print(partner_rate.round(2))
```
